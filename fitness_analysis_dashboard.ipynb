```python
# میرا پہلا Data Analytics کوڈ
print("Data Analytics شروع!")

# نمبروں کا حساب
calories = [409, 479, 340, 282, 406, 300]
print("کل calories:", sum(calories))
print("اوسط calories:", sum(calories)/len(calories))
print("سب سے زیادہ:", max(calories))
print("سب سے کم:", min(calories))
```

```python
import pandas as pd

# آپ کا fitness data
data = {
    'duration': [60,60,60,45,45,60,60,450,30,60],
    'calories': [409,479,340,282,406,300,374,253,195,269],
    'pulse':    [110,117,103,109,117,102,110,104,109,98]
}

df = pd.DataFrame(data)
print("ڈیٹا کی شکل:", df.shape)
print("\nپہلی 5 قطاریں:")
print(df.head())
print("\nاعداد و شمار:")
print(df.describe())
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,4))
plt.bar(range(len(data['calories'])),
        data['calories'],
        color='green')
plt.title('ہر Session میں Calories')
plt.xlabel('Session نمبر')
plt.ylabel('Calories')
plt.show()
```
